```python
import requests
import pandas as pd

url = "https://api.coingecko.com/api/v3/coins/markets"

params = {
    "vs_currency" : "usd",
    "ids" : "bitcoin,ethereum,solana,cardano,dogecoin"
}

headers = {
    "api-key" : "CG-1hWTRmbziAihi68sgvWs5uvj"
}

response = requests.get(
    url,
    params = params,
    headers = headers
)

data = response.json()
df = pd.DataFrame(data)
print(df.head())
```

```python
crypto_df = df[
    ['name','symbol','current_price','total_volume','price_change_percentage_24h']
]
crypto_df.head()
```

```python
from datetime import datetime
crypto_df["timestamp"] = datetime.now()
crypto_df.head()
```

```python
crypto_df.to_csv(
    "crypto_data.csv",
    index=False
    )
print("csv saved")
```
